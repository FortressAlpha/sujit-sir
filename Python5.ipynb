```python
class Student:
    def __init__(self, name, roll):
        self.name = name
        self.roll = roll


    def display(self):
      print("Name:", self.name)
      print("Roll:", self.roll)
s1 = Student("Nidhi", 22)
s1.display()
```

```python
class Rectangle:
  def __init__(self,length,breadth):
    self.length = length
    self.breadth = breadth
  def area(self):
    return self.length*self.breadth

r1 = Rectangle(10,5)
print("Area=", r1.area())
```

```python
class Car:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model
  def display(self):
    print("Brand:", self.brand)
    print("Model:", self.model)
c1 = Car("Toyota", "Camry")
c1.display()
```
