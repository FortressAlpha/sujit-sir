```python
import string

p = input("Enter a paragh:")
words = p.split()
word_counts = {}

for word in words:
  cleaned_word = word.lower().strip(string.punctuation)
  if cleaned_word:
    if cleaned_word in word_counts:
      word_counts[cleaned_word] += 1
    else:
      word_counts[cleaned_word] = 1
print(word_counts)
```
