```python
family = {
    "Rahul":["Amit","Sita"],
    "Riya":["Amit","Sita"],
    "Raj":["Amit","Sita"],
    "Sita":["Mohan","Gita"]
}

def get_parents(person_name):
  """Returns the parents of a given person."""
  return family.get(person_name, [])

def get_children(person_name):
  """Returns the children of a given person."""
  children = []
  for child, parents_list in family.items():
    if person_name in parents_list:
      children.append(child)
  return children

def get_sibling(person_name):
  """Returns the siblings of a given person."""
  siblings = []
  parents_of_person = get_parents(person_name)

  if not parents_of_person:
      return []

  for child_name, child_parents in family.items():
    if child_parents == parents_of_person and child_name != person_name:
      siblings.append(child_name)
  return siblings

person = "Rahul"
print(f"Parents of {person}: {get_parents(person)}")
print(f"Children of {person}: {get_children(person)}")
print(f"Siblings of {person}: {get_sibling(person)}")
```
