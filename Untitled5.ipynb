```python
import numpy as np
import matplotlib.pyplot as plt
def rk2_humble(f,x,y,x_end,h):
  num_step= int((x_end - x)/h)
  x_value = np.zeros(num_step+1)
  y_value = np.zeros(num_step+1)
  x_value[0] = x
  y_value[0] = y

  for i in range(num_step):
    x_i = x_value[i] # Use x_i and y_i for current step values
    y_i = y_value[i]
    k1 = f(x_i, y_i)
    k2 = f(x_i + h, y_i + h * k1)
    y_value[i+1] = y_i + h/2*(k1+k2)
    x_value[i+1] = x_i + h
  return x_value,y_value
```

```python
def my_ods(x,y):
  return x+y
x_initial = 0.0
y_initial = 1.0
x_final = 2.0
step_size = 0.1
x_res,y_res = rk2_humble(my_ods,x_initial,y_initial,x_final,step_size)
```

```python
import numpy as np
import matplotlib.pyplot as plt
def rk2_heun(f,x0,y0,x_end,h):
    num_steps=int((x_end-x0)/h)
    x_values=np.zeros(num_steps+1)
    y_values=np.zeros(num_steps+1)
    x_values[0]=x0
    y_values[0]=y0
    for i in range(num_steps):
        x=x_values[i]
        y=y_values[i]
        k1=f(x,y)
        k2=f(x+h,y+h*k1)
        y_values[i+1]=y+0.5*h*(k1+k2)
        x_values[i+1]=x+h
    return x_values,y_values
def my_ode(x,y):
    return x+y
x_initial=0.0
y_initial=1.0
x_final=2.0
step_size=0.1
x_res,y_res=rk2_heun(my_ode,x_initial,y_initial,x_final,step_size)
print(f"{'x':>6} |{'y(RK2 Approximated)':>20}")
print("-"*31)
for x_val,y_val in zip(x_res,y_res):
    print(f"{x_val:6.2f}|{y_val:20.6f}")
plt.figure(figsize=(8,5))
plt.plot(x_res,y_res,'bo-',label="RK2 Approximation")
y_exact=2*np.exp(x_res)-x_res-1
plt.plot(x_res,y_exact,'r--',label="Exact Analytical Solution")
plt.title("2nd-Order Runge-Kutta Method(Heun's Variant)")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```
